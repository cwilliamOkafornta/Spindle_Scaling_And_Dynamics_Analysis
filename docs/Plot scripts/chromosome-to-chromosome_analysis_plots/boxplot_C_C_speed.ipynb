```python
# library packages
import os
import csv
import glob
import math
import seaborn as sns
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
# input folder 
file_path = "D:/data/Analysis Data/python_analysis/input"
# data_files = os.listdir(file_path)

# output folder
folder_save = "D:/data/Analysis Data/python_analysis/output"
```

```python
with os.scandir(file_path) as data_files:
    for i in data_files:
        
        # read out the files from the folder
        df = pd.read_csv(i, index_col=None)
        
         # make a plot for the first cell stage adding the features needed
        fig, ax = plt.subplots(figsize=(6,6))
        
        # choose the custom color palette 
        palette = ['#006600', '#330066', '#FFAA33', '#3333FF', '#00FFFF', '#b84873']
        # palette = ['#CC2EA5', '#006600', '#330066', '#FFAA33', '#3333FF', '#00FFFF', '#b84873']
        
        # plot the primary data
        ax1 = sns.stripplot(data=df, color='grey', alpha=0.8, jitter=0.1)
        
        # plot features
        '''wild-type'''
        # ax1.axes.set_title('Segregation speed \nChromosome-to-chromosome (Wild-type)', fontsize=15)
        '''ani2-RNA'''
        ax1.axes.set_title('Segregation speed \nChromosome-to-chromosome \nSmall embryo (ani2-RNAi)', fontsize=15)
        
        ax1.set_xlabel("Cell stage",fontsize=15)
        ax1.set_ylabel("Speed [μm/min]",fontsize=15)
        ax1.set(ylim=(0, 35))
        
        # Make the boxplot on the plotted data
        ax2 = sns.boxplot(data=df, width=0.5, palette=palette, ax=ax1)
        
        # save to different folder
        # ax2.figure.savefig(os.path.join(folder_save, 'Speed(wild-type).png'), dpi=300) 
        # ax2.figure.savefig(os.path.join(folder_save, 'Speed(wild-type).svg'))
        
        ax2.figure.savefig(os.path.join(folder_save, 'Speed(ani2_RNAi).png'), dpi=300) 
        ax2.figure.savefig(os.path.join(folder_save, 'Speed(ani2_RNAi).svg'))
        
       
        
      
```
