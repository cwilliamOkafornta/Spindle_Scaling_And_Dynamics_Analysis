### <h1 align="center">**Summary of the spindle dynamics values**</h1>

This script is used for calculating and extracting summary values for the spindle dynamics measurements for all the embryonic conditions. The aim is to determine the mean and the corresponding standard deviation of the measurement of each embryonic cell stage. The resulting outputs are combined to form a new table DataFrame which is subsequently stored as a csv file. The `spindleDynamicsValues` function from `DynamicAnalysisSummary` library is imported to compute for the spindle dynamics summary.

The `spindleDynamicsValues` executes different functions to extract and calculate for different parameters. Each parameter computation has a function which can be executed independently, if need be. To execute `spindleDynamicsValues()`, the defined arguments within the function have to be entered correctly. The description of the arguments are shown below:

**`spindleDynamicsValues`** *`(cell_volume_file_number, chromosome_file_number, poles_file_number, pole_chromosome_file_number, metaphase_pole_file_number, elongation_speed_file_number, segregation_speed_file_number, initial_pole_file_number, folder_input, file_input, save_folder, save_file_name, embryo_condition)`*

`IMPORTANT!` The chronological order of the files in the input folder has be entered numerically as the function's arguments 

* List of files in the input folder: 
    * **cell_volume_file_number**: csv file containing the cell volume
    * **chromosome_file_number**: csv file with the final chromosome-to-chromosome distance measurements
    * **poles_file_number**: csv file with the final pole-to-pole distance measurements 
    * **pole_chromosome_file_number**: csv file with the pole-to-chromosome distance measurements
    * **metaphase_pole_file_number**: csv file with the pole-to-pole distances measurements at metaphase
    * **elongation_speed_file_number**: csv file with the elongation rate measurements of the pole-to-pole distance
    * **segregation_speed_file_number**: csv file with the segregation speed measurements of the chromosome-to-chromosome distance
    * **initial_pole_file_number**: csv file with the initial pole-to-pole distance measurements

* Other function arguments:
    * **folder_input**: Folder with input files
    * **file_input**: Method container with the list of file with the input folder. Example `os.listdir(folder_input)`
    * **save_folder**: Folder to save the output file(s)
    * **save_file_name**: The name of the output file. This should be a `string`. Example: `'Dynamics_files'`
    * **embryo_condition**: The embryonic condition of the computated dynamics summary. It should be a `'string'`

`INPUT FILES`: 
Files from different measured parameters as described in the function arguments above.

`OUTPUT FILES`:
A csv file with a table containing the mean and the corresponding standard deviated of different spindle dynamics parameters.
