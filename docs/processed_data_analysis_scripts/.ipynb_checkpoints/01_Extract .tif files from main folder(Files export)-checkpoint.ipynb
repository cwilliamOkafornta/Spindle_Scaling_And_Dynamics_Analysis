```python
# library packages
import os
import shutil
from skimage.io import imshow, imread, imsave
```

```python
# input root folder path 
main_folder = "D:/data/Analysis Data/Exp_data/Exp_RNAi_c27d9.1/Exp55a_tmr31-c27d9-1-02cell_LLSM_20220118/Exp55a_tmr31-c27d9-1-02cell_LLSM_20220118_Analysis"

# output folder path
save_folder = "D:/data/Analysis Data/Exp_data/Exp_RNAi_c27d9.1/Exp55a_tmr31-c27d9-1-02cell_LLSM_20220118/Exp55a_kymograph/Exp55a_kymo"
```

```python
# read out the file in the first sub-folder 
for subdir1 in os.listdir(main_folder):
    if os.path.isdir(os.path.join(main_folder, subdir1)):
        
        # second sub-folder
        for subdir2 in os.listdir(os.path.join(main_folder, subdir1)):
            if os.path.isdir(os.path.join(main_folder, subdir1, subdir2)):
                
                # third sub-folder
                for subdir3 in os.listdir(os.path.join(main_folder, subdir1, subdir2)):
                    if os.path.isdir(os.path.join(main_folder, subdir1, subdir2, subdir3)) and subdir3.endswith('_kymo'):

                        # files from the second sub-folder
                        for file_name in os.listdir(os.path.join(main_folder, subdir1, subdir2, subdir3)):
                            if file_name.endswith('.tif'):                        
                                file_pathname = os.path.join(main_folder, subdir1, subdir2, subdir3, file_name)

                                # make a new variable for the save folder
                                save_pathname = os.path.join(save_folder, file_name)

                                # make a new save folder it is not existing
                                os.makedirs(os.path.dirname(save_pathname), exist_ok=True)

                                # copy the files to the new save folder
                                shutil.copy2(file_pathname, save_pathname)

```
