# **Statistic t-test script**
### This script is used to call the statistic t-test calculation library to compute for **independent t-test**. The library functions mostly account for different  data population of unequal variance from the dynamic analysis of C. elegans of different conditions, i.e, Wild type and RNAi treated strains.

>  ## **Input data structure**

> ### The input file folder takes in multiple csv files. The identity of each of the csv file of interest can be parsed into the function argument to compute for the statistical test.
> ### The columns of each of the csv file should be structured in the following table header below:
| 1-cell | 2-cell | 4-cell | 8-cell | 16-cell | 32-cell | 64-cell |
| ----------- | ----------- | ----------- | ----------- | ----------- | ----------- | ----------- |
| values | values | values | values | values | values | values |
| values | values | values | values | values | values | values |

### Stated below is the function structure and description for optimize useage;
#### `function(Input_folder, Input_file, Output_folder, Condition_1_Input_file, Condition_2_Input_file, Condition_1_type, Condition_2_type):`
#### `function():` Function name
#### `Input_folder` The folder path variable that contains the input files
#### `Input_file` The variable of the input files within the Input_folder that has been parsed with os.listdir() package
#### `Output_folder` The folder path variable that contains the output files
#### `Condition_1_Input_file` The name of the first input file with the first condition
#### `Condition_2_Input_file` The name of the second input file with the second condition
#### `Condition_1_type` The id of the first condition population data
#### `Condition_1_type` The id of the second condition population data
