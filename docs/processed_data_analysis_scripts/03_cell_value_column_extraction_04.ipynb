```python
''' create a dictionary with dataframes '''
# create a dictionary to add the list of dataframes 
data = {} 
for cell_name in cell_names:
    data[cell_name] = pd.DataFrame()
    
# extract similar columns and add to dataframe of the same dataset
for cell_name in cell_names:
    
    with os.scandir(file_path) as data_files:
        for i in data_files:
            # read out the files from the folder
            df_data = pd.read_csv(i, index_col=None)
            
            # Add the exctracted columns to similar dataframes
            try:                
                file_name = i.name
                col_name = file_name.split('.')[0]
                data[cell_name][col_name] = df_data[cell_name]
            except Exception:
                pass
        
# create a list for the new columns to be added to each table
for cell_name in cell_names:
    list_of_mean = []
    list_of_median = []
    list_of_std = []
    list_of_length = []
    list_of_std_error = []
    
    # assign a new variable name to the table
    df = data[cell_name]
    
    # read through each row of the table and calculate for the mean, std (standard deviation) and SE (Standard Error)
    for row_number in range(len(df)):
        table_row = df.iloc[row_number].dropna().to_list() # drop or ignore all the empty or nan value in each row
        table_row_mean = np.mean(table_row)
        table_row_median = np.median(table_row)
        table_row_std = np.std(table_row)
        table_row_length = len(table_row)
        SE = table_row_std / np.sqrt(table_row_length)
        
        list_of_mean.append(table_row_mean)
        list_of_median.append(table_row_median)
        list_of_std.append(table_row_std)
        list_of_length.append(table_row_length)
        list_of_std_error.append(SE)
    
    # assign new header to the added columns in the table 
    df['mean'] = list_of_mean
    df['median'] = list_of_median
    df['std'] = list_of_std
    df['n'] = list_of_length
    df['SE'] = list_of_std_error
    df['time'] = df_data['time']
    
    # save each table separately as cvs files
    data[cell_name].to_csv(os.path.join(folder_save, cell_name+'.csv'), index=False)
    
```
