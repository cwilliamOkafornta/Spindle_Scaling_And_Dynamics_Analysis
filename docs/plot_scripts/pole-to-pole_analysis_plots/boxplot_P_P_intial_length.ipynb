```python
# library packages
import os
import csv
import glob
import math
import seaborn as sns
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
# input folder
folder = r'D:\data\Analysis Data\python_analysis\input'

# output folder
save_files = r'D:\data\Analysis Data\python_analysis\output'
```

```python
with os.scandir(folder) as data_files:
    for i in data_files:
        
        # read out the files from the folder
        df = pd.read_csv(i, index_col=None)
        
         # make a plot for the first cell stage adding the features needed
        fig, ax = plt.subplots(figsize=(5,5))
        
        # choose the custom color palette 
        # palette = ['#CC2EA5', '#006600', '#330066', '#FFAA33', '#3333FF', '#00FFFF', '#b84873'] # color code with 1-cell stage
        palette = ['#006600', '#330066', '#FFAA33', '#3333FF', '#00FFFF', '#b84873'] # color code without 1-cell stage
        
        # plot the primary data
        ax1 = sns.stripplot(data=df, color='grey', alpha=0.8, jitter=0.1)
        
        # plot features
        # ax1.axes.set_title('Initial length \nPole-to-pole (wild-type)', fontsize=15)
        ax1.axes.set_title('Initial length \nPole-to-pole (Small embryo: ani2-RNAi)', fontsize=15)
        ax1.set_xlabel("Cell stage",fontsize=15)
        ax1.set_ylabel("Distance [μm]",fontsize=15)
        ax1.set(ylim=(0, 15))
        
        # Make the boxplot on the plotted data
        ax2 = sns.boxplot(data=df, width=0.5, palette=palette, ax=ax1)
        
        # save to different folder
        ax2.figure.savefig(os.path.join(save_files, 'Initial length.png'), dpi=300) 
        ax2.figure.savefig(os.path.join(save_files, 'Initial length.svg'))
        
       
        
      
```
