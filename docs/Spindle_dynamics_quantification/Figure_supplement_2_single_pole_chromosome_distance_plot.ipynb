# **<center>Pole-to-pole distance plot as a function of time</center>**

```python
# define a function for the pole-to-pole distance plot as a function of time
def poles_distance_plot_parameter(data, time_col, value_col, color, save_folder, plotfilename,
                   xlim=(-100, 200), ylim=(0, 24),
                   vline_x=0, figsize=(8,8)):
    
    """
    Creates a standardized parameter plot with consistent styling
    
    Parameters:
    data (DataFrame): Input data containing time and measurement columns
    time_col (str): Name of the time column
    value_col (str): Name of the measurement values column
    color (str): Hex color code for the plot line
    xlim/ylim (tuple): Axis limits (default matches original plot)
    vline_x (int): X-position for vertical reference line
    figsize (tuple): Figure dimensions
    """
    sns.set_style(style=None)
    plt.figure(figsize=figsize)
    ax = sns.lineplot(data=data, x=time_col, y=value_col, color=color)
    
    # Axis formatting
    ax.xaxis.set_major_locator(MaxNLocator(integer=True))
    ax.yaxis.set_major_locator(MaxNLocator(integer=True))
    ax.set(xlim=xlim, ylim=ylim)
    
    # Tick customization
    plt.yticks(np.arange(0, 25, 3))
    plt.xticks(np.arange(-100, 220, 30))  # Simplified range specification
    
    # Vertical reference line
    ax.vlines(x=vline_x, ymin=ylim[0], ymax=ylim[1]+3,  # Dynamic ymax based on limits
             linestyle='solid', color='gray', alpha=0.6, linewidth=4)
    
    # Axis labels and tick parameters
    ax.tick_params(axis='both', which='major', labelsize=15, direction='in', size=12, right=True, top=False)
    ax.set_xlabel('Time (s)', fontsize=20)
    ax.set_ylabel('Pole-to-pole distance (µm)', fontsize=25)
    # plt.show()
    
    # Save the plot
    ax.figure.savefig(os.path.join(save_folder, 'WT_pp_'+plotfilename+'.png'), dpi=400, bbox_inches='tight')
    plt.close('all')
```

# **<center>Combined plots for pole-to-pole distance as a function of time</center>**

```python
# define a function for the pole-to-pole distance plot as a function of time
def combined_poles_distance_plot_parameter(WT_data, WT_time_col, WT_value_col, 
                                           small_data, small_time_col, small_value_col,
                                           big_data, big_time_col, big_value_col,
                                           save_folder, plotfilename, 
                                           xlim=(-100, 200), ylim=(0, 24), vline_x=0, figsize=(8,8)):
    
    """
    Creates a standardized parameter plot with consistent styling
    
    Parameters:
    data (DataFrame): Input data containing time and measurement columns
    time_col (str): Name of the time column
    value_col (str): Name of the measurement values column
    color (str): Hex color code for the plot line
    xlim/ylim (tuple): Axis limits (default matches original plot)
    vline_x (int): X-position for vertical reference line
    figsize (tuple): Figure dimensions
    """
    sns.set_style(style=None)
    plt.figure(figsize=figsize)
    ax = sns.lineplot(data=WT_data, x=WT_time_col, y=WT_value_col, color='#004C99')
    sns.lineplot(data=small_data, x=small_time_col, y=small_value_col, color='#009999', ax=ax)
    sns.lineplot(data=big_data, x=big_time_col, y=big_value_col, color='#CC6600', ax=ax)
    
    # Axis formatting
    ax.xaxis.set_major_locator(MaxNLocator(integer=True))
    ax.yaxis.set_major_locator(MaxNLocator(integer=True))
    ax.set(xlim=xlim, ylim=ylim)
    
    # Tick customization
    plt.yticks(np.arange(0, 25, 3))
    plt.xticks(np.arange(-100, 220, 30))  # Simplified range specification
    
    # Vertical reference line
    ax.vlines(x=vline_x, ymin=ylim[0], ymax=ylim[1]+3,  # Dynamic ymax based on limits
             linestyle='solid', color='gray', alpha=0.6, linewidth=4)
    
    # Axis labels and tick parameters
    ax.tick_params(axis='both', which='major', labelsize=15, direction='in', size=12, right=True, top=False)
    ax.set_xlabel('Time (s)', fontsize=20)
    ax.set_ylabel('Distance (µm)', fontsize=25)
    plt.show()
    
    # Save the plot
    ax.figure.savefig(os.path.join(save_folder, 'combined_pp_'+plotfilename+'.png'), dpi=400, bbox_inches='tight')
    plt.close('all')
```
