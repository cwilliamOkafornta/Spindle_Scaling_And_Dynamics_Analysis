```python
import os
import numpy as np 
import pandas as pd
from scipy import stats
```

```python
# input folder 
file_path = 'D:/data/Analysis Data/python_analysis/input'
data_files = os.listdir(file_path)

# output folder
# folder_save = 'D:/data/Analysis Data/python_analysis/output'
print(data_files)
```

```python
file_list = []
for file in data_files:
    filepath = os.path.join(file_path, file)
    print(filepath)
    try:
        df = pd.read_csv(filepath, index_col=None, encoding='utf-8')
    except:
        df = pd.read_csv(filepath, index_col=None, encoding='latin-1')
        
    file_list.append(df)
```

```python
def ttestPoletoPole(folder_input, filedata_input, name_wt_fileinput, name_rnai_fileinput):
    
    WT_final_pp = []
    RNAi_final_pp = []
    
    for file in filedata_input:
        filepath = os.path.join(folder_input, file)
        filename = os.path.basename(filepath)
        if filename == name_wt_fileinput:
            try:
                df = pd.read_csv(filepath, index_col=None, encoding='utf-8')
            except:
                df = pd.read_csv(filepath, index_col=None, encoding='latin-1')

            WT_final_pp.append(df)
            
        if filename == name_rnai_fileinput:
            try:
                df1 = pd.read_csv(filepath, index_col=None, encoding='utf-8')
            except:
                df1 = pd.read_csv(filepath, index_col=None, encoding='latin-1')
            
            RNAi_final_pp.append(df1)
            
#     return WT_final_pp, RNAi_final_pp

    if WT_final_pp:
        new_list = pd.concat(WT_final_pp, ignore_index=True)
        return new_list
    
    # print(WT_final_pp)
    
    return None
    
wt_data= ttestPoletoPole(file_path, data_files, 'Final pole_pole length_Individual (wild-type).csv', 'Final pole_pole length_Individual (ani2-RNAi).csv')
# if wt_data is not None:
#     print(wt_data['1-cell'])
print(wt_data['1-cell'])
```

```python
df_RNAi = file_list[0]
df_WT = file_list[1]
```

```python
cell_2R = (df_RNAi.loc[:,['2-cell']]).dropna()
cell_2W = (df_WT.loc[:,['2-cell']]).dropna()
n_R = len(cell_2R)
n_W = len(cell_2W)
```

```python
mean1, mean2 = np.mean(cell_2R), np.mean(cell_2W)
se1 = stats.sem(cell_2R)
se2 = stats.sem(cell_2W)
```

```python
sed = np.sqrt(se1**2.0 + se2**2)
```

```python
t_stat = (mean1 - mean2) / sed
t_stat
```

```python
deg_of_freedom = n_R + n_W - 2
deg_of_freedom
```

```python
alpha = 0.05
critical_value = stats.t.ppf(1.0 - alpha, deg_of_freedom)
```

```python
critical_value
```

```python
p_value = (1.0 - stats.t.cdf(abs(t_stat), deg_of_freedom)) * 2.0
p_value
```

```python
stat, p= stats.ttest_ind(cell_2W, cell_2R, equal_var=False)
stat, p
```
