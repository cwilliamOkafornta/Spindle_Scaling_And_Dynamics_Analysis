# `First derivative`
### for **pole-to-pole**, **pole-to-chromosome** and **chromosome-to-chromosome**

```python
import pandas as pd
import numpy as np
import os
import math
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# input folder 
file_path = "D:/data/Analysis Data/python_analysis/input"
# data_files = os.listdir(file_path)

# output folder
folder_save = "D:/data/Analysis Data/python_analysis/output"
```

```python
# read into the path directory of the files
with os.scandir(file_path) as data_files:
    for i in data_files:
        
        if i.name.startswith('mean_'):
        # if i.name.startswith('mean_ani2_C_C'):
        # if i.name.startswith('mean_ani2_AVG_P_C'):
        
            # read out the files from the folder
            df = pd.read_csv(i, index_col=None)
            
            # Get the column headers
            headers = df.columns[1:]
            
            # Extract the time and value columns from the DataFrame
            time = df["time"].values
            
            # define a dictionary for the values derivatives
            derivatives = {}
            
            # Find the index where time == 0; use either line below
            start_index = np.argmax(time == 0)
            # start_index = np.where(time == 0)[0][0]
            
            # Calculate the first derivatives from time 0 for each column
            for header in headers:
                if header != "Time":
                    values = df[header].values[start_index:]
                    derivative = np.gradient(values, time[start_index:])
                    derivatives[header] = derivative
                    
            # Create a new DataFrame with the calculated derivatives
            derivative_table = pd.DataFrame({'Time': time[start_index:], **derivatives})
            
            # save the table in a csv file 
            file_path = os.path.join(folder_save, 'P_P_derivatives.csv')
            # file_path = os.path.join(folder_save, 'P_C_derivatives.csv')
            # file_path = os.path.join(folder_save, 'C_C_derivatives.csv')
            # file_path = os.path.join(folder_save, 'P_C_VL_derivatives.csv')
            derivative_table.to_csv(file_path, index=False)
            
```
