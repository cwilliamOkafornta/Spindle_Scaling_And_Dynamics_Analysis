# Pole-to-chromosome normalization plot 
## Ratio: `Final length / Initial length`.
### The goal of the plot is to determine the relative distance over different cell stages by normalizing the pole-to-chromosome final distance to the initial distance.
### `Input files`: csv files containing the mean pole-to-chromosome distances and the individual cell type pole-to-chromomsome distances.
### `Output files`: png and svg files showing the plot visualization.

```python
# library packages
import os
import csv
import glob
import math
import seaborn as sns
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
# input folder 
file_path = "D:/data/Analysis Data/python_analysis/input"

# output folder
folder_save = "D:/data/Analysis Data/python_analysis/output"
```

```python
files = []
with os.scandir(file_path) as data_files:
    for i in data_files:
        # read out the files from the folder
        df = pd.read_csv(i, index_col=None)
        files.append(df)

# assign variables to the items in the list
df1 = files[0]
df2 = files[1]

# group the dataframes as 2 groups
df1['Plot'] = 'Wild-type'
df2['Plot'] = 'Small_embryo'

# merge the 2 dataframes from the list
new_df = pd.concat([df1, df2])

# make a plot for the first cell stage adding the features needed
fig, ax = plt.subplots(figsize=(6,7))

# plot the primary data
ax1 = sns.stripplot(x='variable', y='value', hue='Plot', 
                    data=pd.melt(new_df, id_vars=['Plot']), 
                    size=4, 
                    palette='Accent_r', 
                    jitter=0.2, 
                    dodge=6, 
                    alpha=0.7)

# Make the boxplot on the plotted data
ax2 = sns.boxplot(x='variable', y='value', hue='Plot', data=pd.melt(new_df, id_vars=['Plot']), palette='bone_r', dodge=10)

# plot features
'''Rate of P-C reduction'''
# ax2.axes.set_title('P-C \nRate of reduction \nWT vs. Small embryo', fontsize=15)

'''P-P Elongation rate'''
# ax2.axes.set_title('P-P Elongation rate \nWT vs. Small embryo', fontsize=15)

'''C-C Segregation rate'''
ax2.axes.set_title('C-C Segregation rate \nWT vs. Small embryo', fontsize=15)

# x- and y-axis 
ax2.set_xlabel("Cell stage",fontsize=15)
ax2.set_ylabel("Distance[µm]",fontsize=15)
# ax2.set(ylim=(0, 10))
        
# save to different folder
# ax2.figure.savefig(os.path.join(folder_save, 'Speed_P-C_wt-small_embryo.png'), dpi=300) 
# ax2.figure.savefig(os.path.join(folder_save, 'Speed_P-C_wt-small_embryo.svg'))

# ax2.figure.savefig(os.path.join(folder_save, 'Speed_P-P_wt-small_embryo.png'), dpi=300) 
# ax2.figure.savefig(os.path.join(folder_save, 'Speed_P-P_wt-small_embryo.svg'))

ax2.figure.savefig(os.path.join(folder_save, 'Speed_C-C_wt-small_embryo.png'), dpi=300) 
ax2.figure.savefig(os.path.join(folder_save, 'Speed_C-C_wt-small_embryo.svg'))
```
